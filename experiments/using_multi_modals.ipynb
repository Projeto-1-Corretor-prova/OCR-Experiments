Definindo Constantes

```python
MULTI_MODAL_MODEL_NAME = "gemini-2.5-flash"
```

```python
DIRETORIO_DATA = "data"
```

```python
SYSTEM_PROMPT = (
    "Você é um assistente de OCR especializado em transcrever documentos de avaliação. "
    "Sua tarefa é transcrever o conteúdo do documento de avaliação fornecido, mantendo a estrutura, quebras de linha e formatação original. "
    "Não adicione comentários extras ou interpretações, apenas transcreva o texto exatamente como está no documento."
)
```

```python
HUMAN_PROMPT = (
    "Transcreva este documento de avaliação exatamente como está escrito. "
)
```

Definindo funções

```python
def encode_image(image: Image) -> str:
    buffered = BytesIO()
    image.save(buffered, format="JPEG")
    return b64encode(buffered.getvalue()).decode('utf-8')
```

```python
def process_document(image_path: str, chat: BaseChatModel) -> str:
    image = open(image_path)
    encoded_image = encode_image(image)
    system_message = HumanMessage(content=SYSTEM_PROMPT)
    message = HumanMessage(
        content=[
            {"type": "text", "text": HUMAN_PROMPT},
            {
                "type": "image_url",
                "image_url": f"data:image/jpeg;base64,{encoded_image}"
            },
        ]
    )
    response = chat.invoke([
        system_message,
        message])
    return response.content
```

```python
def process_test(matricula: str, chat: BaseChatModel) -> str:
    image_paths = listdir(f"{DIRETORIO_DATA}/{matricula}")
    
    transcription = str()

    for image_path in image_paths:
        transcription += process_document(f"{DIRETORIO_DATA}/{matricula}/{image_path}", chat)
        transcription += "\n\n"  
        
    return transcription
```
