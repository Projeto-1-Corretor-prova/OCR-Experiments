## OCR

```python
REGEX_QUESTION = r"Resposta [0-9]+\)"
REGEX_ID = r"[0-9]+"
```

```python
OCR_TECHINIQUES: list[tuple[str, OCRInterface]] = [
    ("testando", OCRManagerV1()),
]
```

```python
for technique_name, _ in OCR_TECHINIQUES:
    if not exists(f"{DATA_OUTPUT}/{technique_name}"): 
        mkdir(f"{DATA_OUTPUT}/{technique_name}")
```

```python
for technique_name, technique in OCR_TECHINIQUES:
    final_path = Path(f"{DATA_OUTPUT}/{technique_name}.json")
    ocr_result = technique.ocr({"ocr_path": DATA_INPUT, "regex_id": REGEX_ID, "regex_question": REGEX_QUESTION})
    with open(f"{final_path}", 'w') as f:
        dump(ocr_result, f)
```

```python
for technique_name, _ in OCR_TECHINIQUES:
    with open(f"{DATA_OUTPUT}/{technique_name}.json", 'r') as f:
        print("---")
        print("Technique:", technique_name)
        print(dumps(load(f), indent=4))
        print("---")
```
