## Configurations


```python
DATA_INPUT = "data/output-preprocess/basic-pre-process-technique"

DATA_OUTPUT = "data/output-ocr"
```

```python
if not exists(DATA_INPUT) or not any(listdir(DATA_INPUT)): 
    raise Exception("Input data don't exist")

if not exists(DATA_OUTPUT):
    mkdir(DATA_OUTPUT)
```

## OCR

```python
IMAGE_PATHS = listdir(f"{DATA_INPUT}")
```

```python
OCR_TECHINIQUES: list[tuple[str, OCRInterface]] = [
    ("basic-ocr-technique", OCRManagerV0()),
]
```

```python
for technique_name, _ in OCR_TECHINIQUES:
    if not exists(f"{DATA_OUTPUT}/{technique_name}"): 
        mkdir(f"{DATA_OUTPUT}/{technique_name}")
```

```python
for technique_name, technique in OCR_TECHINIQUES:
    for image_path in tqdm(IMAGE_PATHS, desc=f"Technique: {technique_name}", total=len(IMAGE_PATHS)):
        final_path = Path(image_path).with_suffix(".json")
        ocr_result = technique.ocr(f"{DATA_INPUT}/{image_path}")
        with open(f"{DATA_OUTPUT}/{technique_name}/{final_path}", 'w') as f:
            dump(ocr_result, f)
```

first ocr from every technique

```python
for technique_name, _ in OCR_TECHINIQUES:
    ocr_results = listdir(f"{DATA_OUTPUT}/{technique_name}")
    with open(f"{DATA_OUTPUT}/{technique_name}/{ocr_results[0]}", 'r') as f:
        print("---")
        print("Technique:", technique_name)
        print(dumps(load(f), indent=4))
        print("---")
```
